# Normalisasi Data
 
## Pengertian Normalisasi

Normalisasi adalah teknik pra-pemrosesan data (data preprocessing) yaitu proses mengubah nilai suatu atribut numerik ke dalam rentang atau skala tertentu tanpa mengubah informasi dasar dari data tersebut. Tujuan utama normalisasi adalah agar setiap atribut memiliki skala yang sebanding, sehingga tidak ada atribut yang mendominasi perhitungan.

## Macam - Macam Normalisasi

Beberapa metode normalisasi yang umum digunakan:

### 1. Normalisasi Min-Max

Min-Max Normalization merupakan teknik normalisasi yang mengubah nilai suatu atribut ke dalam rentang tertentu yang telah ditentukan sebelumnya.

Rumus Normalisasi Min-Max:

$$v' = \frac{v - min_A}{max_A - min_A} (new_{max_A} - new_{min_A}) + new_{min_A}$$

Keterangan :

$v$ : nilai asli dari atribut $A$

$v'$ : nilai hasil normalisasi

$min_A$ : nilai minimum dari atribut $A$

$max_A$ : nilai maksimum dari atribut $A$

$new_{min_A}$ : batas minimum rentang normalisasi yang baru

$new_{max_A}$ : batas maksimum rentang normalisasi yang baru

Min-Max Normalization merupakan metode normalisasi yang digunakan untuk mengubah nilai suatu atribut ke dalam rentang tertentu yang telah ditentukan. Proses ini dilakukan dengan memetakan nilai dari rentang awal $[min_A, max_A]$ ke rentang baru $[new_{min_A}, new_{max_A}]$. Dengan menggunakan metode ini, setiap atribut dalam dataset akan memiliki skala yang sama sehingga tidak ada atribut yang memiliki pengaruh lebih dominan dalam proses analisis data.

#### Kasus Normalisasi $[0,1]$

Jika rentang normalisasi yang digunakan adalah $[0,1]$, maka:

Rumus normalisasi $[0,1]$:

$$v' = \frac{v - min_A}{max_A - min_A}$$

Keterangan :

$v$ : nilai asli dari atribut $A$

$v'$ : nilai hasil normalisasi

$min_A$ : nilai minimum dari atribut $A$

$max_A$ : nilai maksimum dari atribut $A$

#### Kasus Normalisasi Selain Rentang $[0,1]$

Normalisasi tidak selalu menggunakan rentang $[0,1]$. Dalam beberapa kasus, data dapat dinormalisasi ke rentang lain sesuai kebutuhan analisis, misalnya $[-1,1]$, $[0,100]$, atau $[1,10]$. Dengan menggunakan rumus umum Min-Max Normalization, nilai data dari rentang awal $[min_A, max_A]$ dapat dipetakan ke rentang baru $[new_min_A, new_max_A]$.

Rumus Normalisasi Selain Rentang $[0,1]$:

$$v' = \frac{v - min_A}{max_A - min_A} (new_{max_A} - new_{min_A}) + new_{min_A}$$

Contoh Normalisasi ke Rentang $[1,10]$

$$new_{min_A} = 1$$

$$new_{max_A} = 10$$

Sehingga rumusnya menjadi:

$$v' = \frac{v - min_A}{max_A - min_A} (10 - 1) + 1$$

$$v' = \frac{v - min_A}{max_A - min_A} (5 - 1) + 1$$

### 2. Normalisasi Z-Score

Dalam beberapa kasus, metode Min-Max Normalization tidak dapat digunakan secara efektif. Hal ini terjadi ketika nilai minimum dan maksimum dari suatu atribut tidak diketahui, atau ketika data mengandung outlier yang dapat mempengaruhi hasil normalisasi. Outlier dapat menyebabkan nilai-nilai data terkonsentrasi pada rentang tertentu sehingga mengurangi ketelitian representasi data.

Sebagai alternatif, digunakan Z-Score Normalization yang menormalisasi nilai data berdasarkan rata-rata (mean) dan standar deviasi dari atribut tersebut. Metode ini mengubah nilai atribut sehingga memiliki rata-rata 0 dan standar deviasi 1.

Rumus Normalisasi Z-Score:

$$v' = \frac{v - \mu_A}{\sigma_A}$$

Keterangan:

$v$ : nilai asli dari atribut $A$

$v'$ : nilai hasil normalisasi

$\mu_A$ : nilai rata-rata (mean) dari atribut $A$

$\sigma_A$ : standar deviasi dari atribut $A$

**Perhitungan Mean (Rata - Rata)**

Jika nilai mean belum diketahui, maka dapat dihitung menggunakan rumus berikut:

$$\mu_A = \frac{1}{n} \sum_{i=1}^{n} v_i$$

Keterangan:

$v_i$ : nilai data ke-$i$

$n$ : jumlah data

$\mu_A$ : rata-rata atribut $A$

**Perhitungan Standar Deviasi**

Standar deviasi dapat dihitung menggunakan rumus berikut:

$$\sigma_A = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(v_i - \mu_A)^2}$$

Keterangan:

$v_i$ : nilai data ke-$i$

$\mu_A$ : rata-rata atribut

$n$ : jumlah data

$\sigma_A$ : standar deviasi atribut $A$

#### Variasi Z-Score Menggunakan Mean Absolute Deviation

Variasi lain dari Z-Score Normalization menggunakan Mean Absolute Deviation (MAD) sebagai pengganti standar deviasi. Pendekatan ini lebih robust terhadap outlier, karena tidak menggunakan kuadrat selisih nilai terhadap rata-rata.

Rumus Mean Absolute :

$$s_A = \frac{1}{n}\sum_{i=1}^{n}|v_i - \mu_A|$$

Keterangan:

$s_A$ : mean absolute deviation dari atribut $A$

$v_i$ : nilai data ke-$i$

$\mu_A$ : rata-rata (mean) dari atribut $A$

$n$ : jumlah data

$|v_i - \mu_A|$ : nilai absolut selisih antara data ke-$i$ dan rata-rata

Rumus Z-Score Normalization Menggunakan Mean Absolute Deviation:

$$v' = \frac{v - \mu_A}{s_A}$$

Keterangan :

$v$ : nilai asli dari atribut $A$

$v'$ : nilai hasil normalisasi

$\mu_A$ : rata-rata (mean) dari atribut $A$

$s_A$ : mean absolute deviation dari atribut $A$

### 3. Normalisasi Decimal Scalling

Decimal Scaling Normalization adalah metode normalisasi data yang dilakukan dengan cara menggeser titik desimal dari nilai atribut sehingga seluruh nilai berada dalam rentang tertentu (biasanya antara −1 sampai 1).

Normalisasi ini dilakukan dengan membagi setiap nilai data dengan pangkat 10 tertentu. Nilai pangkat tersebut ditentukan berdasarkan nilai absolut maksimum dari atribut sehingga hasil normalisasi memiliki nilai kurang dari 1. Metode ini sederhana karena hanya menggunakan operasi pembagian dengan $10^j$.

Rumus Normalisasi Decimal Scaling:

$$v' = \frac{v}{10^j}$$

Keterangan:

$v$ : nilai asli dari atribut

$v'$ : nilai hasil normalisasi

$j$ : jumlah digit dari nilai absolut terbesar pada atribut 

$10^j$ : faktor skala untuk menggeser titik desimal

**Cara Menentukan Nilai j**

1. Cari nilai terbesar (absolute) pada atribut

2. Hitung jumlah digitnya

3. Jumlah digit tersebut menjadi nilai $j$

Metode Decimal Scaling Normalization memiliki kelebihan karena sangat sederhana dan cepat dihitung, terutama pada dataset yang memiliki skala nilai besar. Namun, metode ini kurang fleksibel dibandingkan Min-Max atau Z-Score, karena hanya bergantung pada jumlah digit nilai maksimum dan tidak mempertimbangkan distribusi data.

Rumus Normalisasi Decimal Scaling:

$$v' = \frac{v}{10^j}$$

Sebelum melakukan perhitungan normalisasi menggunakan metode Decimal Scaling harus mengetahui nilai $j$ terlebih dahulu. Cara mengetahui nilai $j$ yaitu dengan memilih nilai terbesar pada atribut lalu menghitung jumlah digitnya.

**Menentukan nilai $j$ pada atribut IPK**

Dari data yang akan di normalisasi pada atribut IPK, nilai terbesarnya adalah 4. Nilai 4 ini hanya 1 digit, jadi $j$ pada atribut IPK = $1$.

**Menentukan nilai $j$ pada atribut PO**

Dari data yang akan di normalisasi pada atribut PO, nilai terbesarnya adalah 400000. Nilai 400000 ini berjumlah 6 digit, jadi $j$ pada atribut PO = 6$.

**Perhitungan Normalisasi Decimal Scaling Pada Atribut IPK**

$$v' = \frac{v}{10^j}$$

$$v' = \frac{2}{10^1}$$

$$v' = 0.2$$

Jadi, diperoleh nilai $0.2$ dari hasil normalisasi pada atribut **IPK** data ke - 1.

```python
# Decimal Scalling

import numpy as np

def decimal_scaling(data):
    max_val = np.max(np.abs(data))
    j = len(str(int(max_val)))
    return data / (10**j)

data = np.array([[2],[3],[4], [2], [3], [4], [2]])

hasil = decimal_scaling(data)

print(hasil)
```

Data tersebut berjumlah 7 data dan memiliki 3 fitur. Tetapi pada Fitur **JML** data ke - 7 terdapat missing value. Ini adalah hal yang harus diselesaikan menggunakan metode **WKNN**.

Langkah - Langkah Penyelesaian:

**1. Normalisasi Data** 

    Langkah pertama yang harus dilakukan untuk menyelesaikan missing value ini adalah menormalisasikan data supaya data berada di rentang $0 - 1$. Di sini normalisasi menggunakan $min-max$.

    Rumus Normalisasi :

    $$x' = \frac{x_i - x_{min}}{x_{max} - x_{min}}$$

    Berikut hasil Normalisasi Data:

    | no | ipk_norm | po_norm | jml_norm |
    |----|----------|---------|----------|
    | 1  | 0        | 0       | 0        |

    Setelah dilakukan normalisasi, tahap selanjutnya adalah menghitung jarak dan kemiripan.

**2. Hitung Jarak dan Kemiripan**

    Pada data ini terdapat 2 fitur yang digunakan untuk perhitungan jarak untuk mengisi missing value. Fiturnya adalah **ipk** dan **po**. Kemudian di hitung kemiripan jarak dengan rumus kemiripan $s_i = \frac {1}{distance}$.

    Untuk menghitung distance menggunakan rumus variasi dari Jarak Euclidean:

    $$\frac{1}{s_i} = \sum_{h_i \in O_i \cap O_j} (y_{ih} - y_{jh})^2$$

    Keterangan :

    $y_i$ = data ke-i

    $y_j$ = data tetangga
    
    $O_i \cap O_j$ = atribut yang sama-sama tersedia

    $s_i$ = nilai kemiripan

    Berikut merupakan hasil setelah perhitungan distance dan simmilarity dari data yang telah di normalisasi:

    | $y_i$ | $(y_{i,ipk}-y_{j,ipk})^2$| $(y_{i,po}-y_{j,po})^2$| $\sum_h (y_{ih}-y_{jh})^2$| $s_i=\frac{1}{\sum}$ |
    | :---: | :---------------:        | :--------------:    | :----------------:| :------------------: |
    | $y_1$ | $(0 - 0)^2 = 0$          | $(0.5 - 0)^2 = 0.25$|        0.25       |          4.0         |
    | $y_2$ | $(0 - 0.5)^2 = 0.25$     |  $(0.5 - 0.5)^2 = 0$|        0.25       |          4.0         |
    | $y_3$ | $(0 - 1)^2 = 1$          | $(0.5 - 0)^2 = 0.25$|        1.25       |          0.8         |
    | $y_4$ | $(0 - 0)^2 = 0$          | $(0.5 - 0)^2 = 0.25$|        0.25       |          4.0         |
    | $y_5$ | $(0 - 0.5)^2 = 0.25$     | $(0.5 - 0.5)^2 = 0$ |        0.25       |          4.0         |
    | $y_6$ | $(0 - 1)^2 = 1$          | $(0.5 - 1)^2 = 0.25$|        1.25       |          0.8         |

    Setelah nilai simmilarity/kemiripan sudah didapatkan, langkah selanjutnya adalah menghitung estimasi menggunakan Weighted Average.

**3. Hitung Estimasi Menggunakan Weighted Average**

   Untuk menghitung ini adalah dengan cara mengkalikan setiap atribut **JML** $(y_i)$ dengan **Kemiripan** $(s_i)$, kemudian di bagi dengan Total Kemiripan.

   A. Pembilang $\sum (s_i * y_{ih})$ :

      $y_1 : 4.0 * 0 = 0$

      $y_2 : 4.0 * 1 = 4$

      $y_3 : 0.8 * 0 = 0$

      $y_4 : 4.0 * 1 = 4$

      $y_5 : 4.0 * 0 = 0$

      $y_6 : 0.8 * 1 = 0.8$

      Total Pembilang:

        $0 + 4 + 0 + 4 + 0 + 0.8 = 8.8$

   B. Penyebut $(\sum s_i)$

        $4.0 + 4.0 + 0.8 + 4.0 + 4.0 + 0.8 = 17.6$


**4. Hasil Akhir**

   Setelah menghitung pembilang dan penyebut, langkah terakhir yaitu menghitung hasil akhir dengan rumus:

   $y = \frac {Pembilang} {Penyebut}$

   $y = \frac {8.8} {17.6} = 0.5$

   **Kesimpulan**

    Berdasarkan perhitungan menggunakan metode **Weighted K-Nearest Neighbor (WKNN)**, nilai estimasi untuk atribut **JML** pada data ke-7 yang sebelumnya memiliki missing value berhasil diperoleh melalui perhitungan **weighted average** dari nilai kemiripan setiap data tetangga.
    
    Hasil perhitungan menunjukkan bahwa nilai estimasi yang diperoleh adalah **0.5** pada skala data yang telah dinormalisasi. Nilai tersebut merupakan hasil dari pembagian antara total hasil perkalian similarity dengan nilai **JML** terhadap total nilai similarity.
    
    Dengan demikian, missing value pada fitur **JML** untuk data ke-7 dapat diisi dengan nilai estimasi **0.5** berdasarkan perhitungan menggunakan metode **WKNN**.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsRegressor as KNR

data = {
    'IPK': [2, 3, 4, 2, 3, 4, 2],
    'PO': [2000000, 3000000, 2000000, 2000000, 3000000, 4000000, 3000000],
    'JML': [2, 3, 2, 3, 2, 3, np.nan]
}
df = pd.DataFrame(data)

# Data Training
df_train = df.iloc[:6].copy()

# Data Tes
df_test = df.iloc[6:].copy()

# Normalisasi Min-Max
scaler_features = MinMaxScaler()
scaler_target = MinMaxScaler()

# Normalisasi Fitur (IPK & PO)
train_X = scaler_features.fit_transform(df_train[['IPK', 'PO']])
test_X = scaler_features.transform(df_test[['IPK', 'PO']])

print(train_X)

# Normalisasi Target (JML)
train_y = scaler_target.fit_transform(df_train[['JML']])

# 4. Membangun Model KNN
# Kita gunakan n_neighbors=6 agar semua data latih diperhitungkan sesuai hitungan manual.
# Rumus (1) menyebutkan 1/s = jarak kuadrat. 
# Maka s (bobot) = 1/jarak kuadrat.
def custom_weights(distances):
    return 1 / (distances**2)

knn = KNR(n_neighbors=6, weights=custom_weights, metric='euclidean')
knn.fit(train_X, train_y)

# 5. Prediksi dan Denormalisasi
# Melakukan prediksi pada baris ke-7
pred_scaled = knn.predict(test_X)

# Mengembalikan nilai ke skala asli (Denormalisasi)
pred_final = scaler_target.inverse_transform(pred_scaled)

print(f"Prediksi JML (Ternormalisasi): {pred_scaled[0][0]}")
print(f"Hasil Akhir Prediksi JML: {pred_final[0][0]}")
```
